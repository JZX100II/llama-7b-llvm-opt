```python
import os
import shutil
from glob import glob
from pathlib import Path
from tqdm import tqdm
import subprocess

# Paths
SRC_ROOT = "/content/dataset/ProgramData"

DST_ROOT = "/content/tsvb_dataset_ll"

# Step 1: Create destination directory
os.makedirs(DST_ROOT, exist_ok=True)

# Step 2: Gather all .txt files
train = 1000
txt_files = glob(f"{SRC_ROOT}/**/*.txt", recursive=True)[:train]

print(f"Found {len(txt_files)} files. Starting compilation...")

for path in tqdm(txt_files):
    try:
        # Your existing processing code here
        name = Path(path).stem
        c_path = f"/tmp/{name}.c"
        ll_path = f"{DST_ROOT}/{name}.ll"

        shutil.copy(path, c_path)
        subprocess.run(["clang", "-S", "-emit-llvm", c_path, "-o", ll_path],
                       check=True,
                       stdout=subprocess.DEVNULL,
                       stderr=subprocess.DEVNULL)
    except Exception as e:
        print(f"❌ Failed to compile {path}: {e}")

print("\n✅ All valid .txt files converted to .ll and stored in:", DST_ROOT)
```

```python
from ctransformers import AutoModelForCausalLM
from transformers import LlamaTokenizer
from datasets import load_dataset
from glob import glob
from tqdm import tqdm
import shutil
import subprocess
from pathlib import Path
import pandas as pd

class PassListOptimizer:
    def __init__(self, model_path="legraphista/llm-compiler-7b-IMat-GGUF", quant="Q4_K", dataset_path=None):
        self.model = AutoModelForCausalLM.from_pretrained(
            model_path,
            model_file=f"llm-compiler-7b.{quant}.gguf",
            model_type="llama",
            gpu_layers=50
        )
        # self.tokenizer = LlamaTokenizer.from_pretrained("legraphista/llm-compiler-7b-IMat-GGUF")
        if dataset_path:
            self.dataset = self._load_dataset(dataset_path)
        else:
            self.dataset = None

    def _load_dataset(self, path):
        return load_dataset("json", data_files=path)["train"]

    def __call__(self, ir_path, iterations=3):
        current_ir = ir_path
        for _ in range(iterations):
            pass_list = self.generate_pass_list(current_ir)
            optimized_ir = self.apply_pass_list(current_ir, pass_list)
            current_ir = self.refine_based_on_feedback(current_ir, optimized_ir, pass_list)
        return current_ir

    def generate_pass_list(self, ir_path):
        with open(ir_path, "r") as f:
            ir_code = f.read()

        instruction = "Suggest an LLVM optimization pass list for the following IR:\n\n"
        prompt_text = instruction + ir_code

        input_tokens = self.model.tokenize(prompt_text)
        max_context = 512

        if len(input_tokens) > max_context:
            input_tokens = input_tokens[:max_context]

        # Just call generate with input_tokens only
        output_tokens = self.model.generate(input_tokens)

        output_text = self.model.detokenize(output_tokens)
        return self._parse_pass_list(output_text)

    def _parse_pass_list(self, text):
        return [p.strip() for p in text.split(",") if "pass" in p or "-" in p]

    def apply_pass_list(self, ir_path, pass_list):
        output_file = "optimized.bc"
        opt_command = ["opt", *pass_list, ir_path, "-o", output_file]
        subprocess.run(opt_command, check=True)
        return output_file

    def refine_based_on_feedback(self, old_ir, new_ir, old_pass_list):
        # Placeholder for learning-based refinement
        return new_ir


ll_files = glob("/content/tsvb_dataset_ll/**/*.ll", recursive=True)[:1]

optimizer = PassListOptimizer()

results = []

for f in tqdm(ll_files, desc="Optimizing IR files"):
    try:
        optimized_path = optimizer(f)

        original_bc = f.replace(".ll", ".bc")
        subprocess.run(["llvm-as", f, "-o", original_bc], check=True)

        size_before = Path(original_bc).stat().st_size
        size_after = Path(optimized_path).stat().st_size

        percent_reduced = round((size_before - size_after) / size_before * 100, 2)
        results.append({
            "file": f,
            "size_before": size_before,
            "size_after": size_after,
            "percent_reduced": percent_reduced
        })

        optimized_final_path = f.replace(".ll", "_llmopt.bc")
        shutil.move(optimized_path, optimized_final_path)

    except Exception as e:
        print(f"❌ Error on {f}: {e}")

df = pd.DataFrame(results)
print(df.head())

if not df.empty:
    print(f"\nAverage reduction: {df['percent_reduced'].mean():.2f}%")

df.to_csv("llm_optimizer_results.csv", index=False)
```

```python
from ctransformers import AutoModelForCausalLM
from transformers import LlamaTokenizer
from datasets import load_dataset
from glob import glob
from tqdm import tqdm
import shutil
import subprocess
from pathlib import Path
import pandas as pd

# Load your model (adjust path and params as needed)
quant = "Q4_K"

model = AutoModelForCausalLM.from_pretrained(
    "legraphista/llm-compiler-7b-IMat-GGUF",
    model_file=f"llm-compiler-7b.{quant}.gguf",
    model_type="llama",
    gpu_layers=50
)

# ll_files = glob("/content/tsvb_dataset_ll/**/*.ll", recursive=True)[:1]

# results = []

# for f in tqdm(ll_files, desc="Optimizing IR files"):
#     try:
#         optimized_path = model(f, max_new_tokens=50)
#         print("\nllm responded:")
#         print(optimized_path)
#     except Exception as e:
#         print(f"❌ Error on {f}: {e}")

# === Sample LLVM IR optimization test ===

example_ir = """
; ModuleID = 'example'
source_filename = "example.c"

define i32 @main() {
entry:
  %0 = alloca i32, align 4
  store i32 0, i32* %0, align 4
  %1 = load i32, i32* %0, align 4
  ret i32 %1
}
"""

example_ir = """%3 = alloca i32, align 4"""

# Save the IR snippet to a file
ir_path = Path("example.ll")
ir_path.write_text(example_ir)

# Create prompt for the LLM
prompt = f"Suggest LLVM optimization passes (comma-separated) to optimize this IR:\n\n{example_ir}"

# Generate passes suggested by LLM
passes_text = model(prompt, max_new_tokens=50)
print("\nLLM suggested passes:")
print(passes_text)

# Parse passes list, clean empty strings
pass_list = [p.strip() for p in passes_text.split(",") if p.strip()]

# Build opt command with passes
opt_command = ["opt"] + [f"-{p}" for p in pass_list] + [str(ir_path), "-o", "optimized.bc"]

print("\nRunning opt with command:")
print(" ".join(opt_command))

# Run opt to optimize IR
subprocess.run(opt_command, check=True)

print("\nOptimization done. Output saved to optimized.bc")
```
